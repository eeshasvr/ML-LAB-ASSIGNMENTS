8)Code for column Deletion

```python
import pandas as pd
# Import the pandas library and give it the alias 'pd'

info = {'one': pd.Series([1, 2], index=['a', 'b']),
        # Create the 'one' Series with values and index labels
        'two': pd.Series([1, 2, 3], index=['a', 'b', 'c'])}
        # Create the 'two' Series with values and index labels

df = pd.DataFrame(info)
# Create a DataFrame using the 'info' dictionary

print("The DataFrame:")
# Display a message before printing the DataFrame

print(df)
# Display the original DataFrame

# using del function
# Use the del function to delete a column from the DataFrame

print("Delete the first column:")
# Display a message indicating that the first column will be deleted

del df['one']
# Delete the 'one' column from the DataFrame

print(df)
# Display the DataFrame after deleting the 'one' column

# using pop function
# Use the pop() function to delete another column

print("Delete the another column:")
# Display a message indicating that another column will be deleted

df.pop('two')
# Delete the 'two' column from the DataFrame using the pop() function

print(df)
# Display the final DataFrame after deleting the 'two' column
#roll.no"AP24110011506"
```

13)Code for deletion of rows using index label.If in case,the label is duplicate then multiple rows will be deleted

```python
import pandas as pd
# Import the pandas library and give it the alias 'pd'

a_info = pd.DataFrame([[4, 5], [6, 7]], columns=['x', 'y'])
# Create the first DataFrame with columns x and y

b_info = pd.DataFrame([[8, 9], [10, 11]], columns=['x', 'y'])
# Create the second DataFrame with columns x and y

a_info = pd.concat([a_info, b_info])
# Combine the two DataFrames row-wise

print(a_info)
# Display the combined DataFrame

# Drop rows with label 0
a_info = a_info.drop(0)
# Delete all rows having the index label 0

print("Drop the row with label 0")
# Display a message indicating that rows with label 0 are removed

print(a_info)
# Display the DataFrame after deleting rows with label 0
```

17)null values identification in given set

```python
# Import the pandas library
import pandas as pd

# Import the NumPy library
import numpy as np

# Create a DataFrame with Person names and Age values
# np.nan represents a missing or null value
info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27., 29, np.nan, 32]
})

# Print the original DataFrame
print(info)

# Count the number of non-null values in each column
print(info.count())

# Print a message indicating null values
print("Null values")

# Check each value in the DataFrame
# Returns True if the value is null, otherwise False
print(info.isnull())

# Count the total number of null values in each column
print(info.isnull().sum())
#ROLL NO.AP24110011506
```

18)sorting order

```python
# Import the pandas library
import pandas as pd

# Import the NumPy library
import numpy as np

# Create a DataFrame with Person names and Age values
# np.nan represents a missing or null value
info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27., 29, np.nan, 32]
})

# Print the original DataFrame
print(info)

# Print a message for sorting in ascending index order
print("Sorting Order")

# Sort the DataFrame based on the index in ascending order
print(info.sort_index())

# Print a message for sorting in descending index order
print("Descending Order")

# Sort the DataFrame based on the index in descending order
print(info.sort_index(ascending=False))
#ROLL.NO:AP24110011506
```

19)aggregate functions

```python
import pandas as pd

# Read Iris dataset using the online link
df = pd.read_csv("https://raw.githubusercontent.com/fslaborg/datasets/main/data/iris.csv")

# Print the complete dataset
print(df)

# Calculate mean values for each species
mean_df = df.groupby('species').mean()

print("Mean values for each species:")
print(mean_df)

# Calculate sum of values for each species
sum_df = df.groupby('species').sum()

print("\nSum of values for each species:")
print(sum_df)
```
