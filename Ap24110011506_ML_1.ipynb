### Student Practice

Using the `marks` array above:
1. Extract all marks that are failing (below 40).
2. Count how many students scored above 80 (hint: `.sum()` on a boolean array counts `True` values).
3. Replace all failing marks with 0 using `np.where`.

# Write your answer here


```python
import numpy as np

# Create the marks array
marks = np.array([55, 62, 48, 91, 73])

# 1. Extract all marks that are failing (below 40)
failing_marks = marks[marks < 40]
print("Failing marks:", failing_marks)

# 2. Count how many students scored above 80
# The condition creates a boolean array (True/False)
# .sum() counts the True values
above_80 = (marks > 80).sum()
print("Students scoring above 80:", above_80)

# 3. Replace all failing marks with 0 using np.where
updated_marks = np.where(marks < 40, 0, marks)
print("Marks after replacing failing marks:", updated_marks)
# roll number : Ap24110011506
```

### Student Practice

Pick two sentences of your own (write them below) — one pair that should be *similar* in meaning/topic, and one pair that should be *very different*. Compute the cosine similarity for both pairs and confirm your intuition matches the numbers.


```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Pair 1: Similar sentences
sentence1 = "I love learning Python programming"
sentence2 = "I enjoy studying Python coding"

# Pair 2: Very different sentences
sentence3 = "I love learning Python programming"
sentence4 = "The weather is very cold today"

# Convert the sentences into Bag-of-Words vectors
vectorizer = CountVectorizer()

# Fit on all four sentences so they use the same vocabulary
vectors = vectorizer.fit_transform([sentence1, sentence2, sentence3, sentence4])

# Calculate cosine similarity for the similar pair
similarity_pair1 = cosine_similarity(vectors[0], vectors[1])[0][0]

# Calculate cosine similarity for the different pair
similarity_pair2 = cosine_similarity(vectors[2], vectors[3])[0][0]

# Display the results
print("Pair 1 similarity:", similarity_pair1)
print("Pair 2 similarity:", similarity_pair2)
# roll number : Ap24110011506
```
